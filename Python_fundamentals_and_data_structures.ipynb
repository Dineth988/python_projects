```python
## Set unions. Returns a new set that contains all unique elements from both sets
pet_animals = {"dog", "cat", "hamster", "parrot"}
farm_animals = {"cow", "chicken", "goat", "dog", "cat"}

## Syntax 1
union1 = pet_animals | farm_animals
print(union1)

## Syntax 2
union2 = pet_animals.union(farm_animals)
print(union2)
```

```python
## Set intersection. intersection of two sets is a new set containing only the elements common to both sets.
pet_animals = {"dog", "cat", "hamster", "parrot"}
farm_animals = {"cow", "chicken", "goat", "dog", "cat"}

## Syntax 1
intersection1 = pet_animals & farm_animals
print(intersection1)

## Syntax 2
intersection2 = pet_animals.intersection(farm_animals)
print(intersection2)
```

```python
## Difference between two sets is a new set containing elements that are in the first set but not in the second.
registered_users = {"Alice", "Bob", "Charlie", "Diana", "Linda"}
checked_in_users = {"Alice", "Charlie", "Linda"}

print(registered_users - checked_in_users)
```

```python
## Symmetric difference between two sets is a new set containing all the elements that appear in either set but not both
morning_attendees = {"Alice", "Charlie", "Linda", "John", "Jane"}
afternoon_attendees = {"Charlie", "Linda", "Bob", "Jane"}

## Syntax 1
print(morning_attendees ^ afternoon_attendees)

## Syntax 2
print(morning_attendees.symmetric_difference(afternoon_attendees))
```

```python
## Student records
students_dict = {
    1001: {
        "name": "Ayesha Fernando",
        "age": 18,
        "gender": "Female",
        "grade": "12",
        "section": "A",
        "email": "ayesha.fernando@school.edu",
        "phone": "0771234567",
        "city": "Colombo",
        "gpa": 3.85
    },
    1002: {
        "name": "Kasun Perera",
        "age": 17,
        "gender": "Male",
        "grade": "11",
        "section": "B",
        "email": "kasun.perera@school.edu",
        "phone": "0712345678",
        "city": "Negombo",
        "gpa": 3.42
    },
}
```

```python
def search_students(search_key, search_val):
    print(f"\nSearching by {search_key}: {search_val}")
    if search_key == "id":
        if search_val in students_dict.keys():
            for key, val in students_dict[int(search_val)].items():
                print(f"{key} : {val}")
        else:
            print(f"{search_val} is not a registered !")
    elif search_key == "name":
        found = False
        for key, val in students_dict.items():
            if search_val.lower() in val["name"].lower():
                found = True
                for key, vals in val.items():
                    print(f"{key} : {vals}")
                break
        if found != True:
            print(f"{search_val} is not a registered name!")
    else:
        print(f"{search_key} not supported")


s_key, s_val = input("Enter search key and search val leaving between a space between them.").split()
search_students(s_key, s_val)
```

```python
def filter_by_gender(filter_key, filter_val):
    if filter_key == "gender":
        for key, val in students_dict.items():
            if val["gender"].lower() == filter_val.lower():
                for id, det in val.items():
                    print(f"{id} : {det}")
    elif filter_key == "city":
        for key, val in students_dict.items():
            if val["city"].lower() == filter_val.lower():
                for id, det in val.items():
                    print(f"{id} : {det}")

    else:
         print(f"{filter_key} not supported")
        
filter_by_gender("gender", "Female")
```
