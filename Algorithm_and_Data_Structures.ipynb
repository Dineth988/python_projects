```python
## Implementation of a hash table

hash_list = [[] for _ in range(10)]

def hash_fn(word):
    sum = 0
    for char in word:
        sum += ord(char)
    return sum%10

def add(element):
    index = hash_fn(element)
    hash_list[index].append(element)

def look_up(element):
    index = hash_fn(element)
    if element in hash_list[index]:
        sub_list_ind =  hash_list[index].index(element)
    return [index, sub_list_ind]
        

add("dineth") 
add("Lisa") 
add("Stuart") 
print(look_up("Stuart"))
print(hash_list)
```

```python
## Implementing a liner search
def lin_search(iterable, value):
    for item in iterable:
        if item == value:
            return f"index of {value}: {iterable.index(item)}"
        elif item == iterable[-1]:
            return f"{value} not found"

names = ["din", "den", "lin", "san"]
print(lin_search(names, "li"))
```

```python
## Implementing a binary search
def bin_search(iterable, value):
    sorted_iterable = sorted(iterable)
  
    found = False
    list_end = False
     
    while(found != True and list_end != True):
        if len(sorted_iterable) != 0:
            list_end = True
        else:
            if value == sorted_iterable[len(sorted_iterable)//2]:
                found =True
                return f"index: {iterable.index(value)}"
            elif value < sorted_iterable[len(sorted_iterable)//2]:
                if len(sorted_iterable) > 1:
                    sorted_iterable = sorted_iterable[:len(sorted_iterable)//2]    
                else:
                    list_end = True
            elif value > sorted_iterable[len(sorted_iterable)//2]:
                if len(sorted_iterable) > 1:
                    sorted_iterable = sorted_iterable[len(sorted_iterable)//2 + 1:]    
                else:
                    list_end = True
    return f"{value} is not in {iterable}"
la = [1,2,3,4,5]
print(bin_search(la, 7))
```

```python
## Implemanting bubble sort
def bubble_sort(itrb):
    itrb = itrb
    for rnd in range(len(itrb)-1):
        for num in range(len(itrb)-rnd-1):
            if itrb[-1] != itrb[num]:
                if itrb[num] > itrb[num+1]:
                    itrb[num+1], itrb[num] = itrb[num], itrb[num+1]
     
    return itrb
bubble_sort([7,6,4,3,5])   
```

```python
## Implementation of selection sort

def insrt_sort(itrb):
    sorted_list = []
    n = len(itrb)
    for pos in range(n):
        min_num = min(itrb)
        sorted_list.append(min_num)
        itrb.remove(min_num)
    return sorted_list
    
l1 = [7,5,6,2]
insrt_sort(l1)
     
```
