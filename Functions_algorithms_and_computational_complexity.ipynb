```python
## Default values for parameters
def print_name(name = "dineth"):
    print(name)

print_name()
```

```python
## Arbitrary Arguments - *args
def example(*name):
    print(name)

example("dineth", 1, True)
```

```python
## Sum with *args. this create a tuple of arguments
def sum_num(*nums):
    print(sum(nums))

## Max with *args
def max_num(*nums):
    print(max(nums))

sum_num(1,2,4,56,6)
max_num(2,4,9,1)
```

```python
## Arbitrary Keyword Arguments - **kwargs. this create a dictionary of arguments
def name_fun(**names):
    for key, name in names.items():
        print(f"{key} : {name}")

name_fun(fname = "dineth", lname = "denethi")
```

```python
## using * and ** for unpacking into arguments
def print_name(lname, fname, mname):
    print(lname, fname, mname)

dineth = ["wik", "dineth", "dilmin"]
print_name(*dineth)


def my_function(fname, lname):
  print("Hello", fname, lname)

person = {"fname": "Emil", "lname": "Refsnes"}
my_function(**person)
```

```python
import time   
## Decorators
def decorator_func(func):
    def enhanced_fn(*args):
        start_tm = time.time()
        results = func(*args)
        end_tm = time.time()
        print(end_tm - start_tm)
        return results
    return enhanced_fn

@decorator_func
def print_name(name1, name2):
    print(name1, name2)

@decorator_func
def print_names():
    names = ["dineth", "dene", "ima", "dinto"]
    for name in names:
        print(name)
        
print_name("dineth", "dene")
print_names()
```

```python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
```

```python
import functools

def changecase(func):
  @functools.wraps(func)
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

print(myfunction.__name__)
```
