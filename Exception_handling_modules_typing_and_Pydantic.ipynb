```python
## Creating a custom exception

class PlatformNotFound(Exception):
    """Incompatible platform."""
    
def linux_interaction():
    import sys
    if "linux" not in sys.platform:
        raise PlatformNotFound("Function can only run on Linux systems.")
    print("Doing Linux things.")

try:
    linux_interaction()
except PlatformNotFound as e:
    print(e)
    print("linux_interaction function never ran")
else:
    print("else ran")

finally:
    print("finally")
```

```python
from enum import Enum
from uuid import UUID, uuid4
from datetime import date
from pydantic import BaseModel, EmailStr, Field

class Department(Enum):
    HR = "HR"
    SALES = "SALES"
    IT = "IT"
    ENGINEERING = "ENGINEERING"

class Employee(BaseModel):
    employee_id: UUID = Field(default_factory = uuid4, frozen = True)
    name: str = Field(min_length = 3)
    email: EmailStr
    date_of_birth: date = Field(alias = "birth_date")
    salary: float = Field(alias = "compensation", gt = 0)
    department: Department
    elected_benefits: bool
```

```python

e1 = Employee(
    name = "Dill",
    email = "dil@gmail.com",
    date_of_birth = "2002-05-08",
    salary = 10.00,
    department = "IT",
    elected_benefits = True,
)
e1
```

```python
from datetime import datetime, date, timedelta
from pydantic import field_validator, BaseModel, EmailStr, Field
from uuid import uuid4, UUID
from enum import Enum
class Department(Enum):
    HR = "HR"
    SALES = "SALES"
    IT = "IT"
    ENGINEERING = "ENGINEERING"

class Employee(BaseModel):
    employee_id: UUID = Field(default_factory = uuid4, frozen = True)
    name: str = Field(min_length = 3)
    email: EmailStr
    date_of_birth: date = Field(alias = "birth_date")
    salary: float = Field(alias = "compensation", gt = 0)
    department: Department
    elected_benefits: bool

@field_validator("date_of_birth")
@classmethod
def age_is_18(bdate: date) -> str:
    eigt_years_ago = date(date.today().year - 18, date.today().month, date.today().day)
    if bdate < eigt_years_ago:
        print("You are a adult")
    else:
        print("Minor")
    
```
