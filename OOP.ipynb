```python
## OOP exercises
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    def get_balance(self):
        return self.__balance
    def set_balance(self, new_balance):
        self.__balance = new_balance
        
    def deposit(self, depo):
        self.set_balance(depo+self.__balance)

b1 = BankAccount(100)
print(b1.get_balance())
b1.set_balance(1000)
print(b1.get_balance())
b1.deposit(10)
print(b1.get_balance())
```

```python
## With properties
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
        
    @property    
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, new_balance):
        self.__balance = new_balance

    def deposit(self, depo_amnt):
        self.balance = self.__balance + depo_amnt

b1 = BankAccount(1000)
b1.deposit(100)

```

```python
class Passenger:
    def __init__(self, name):
        self.name = name
       

class Flight:
    def __init__(self, fl_num, capacity):
        self.capacity = capacity
        self.fl_num = fl_num
        self.passengers = []

    def book(self, passenger):
        if self.capacity < len(self.passengers):
            print("No more room")
        else:
            self.passengers.append(passenger)
            print(f"{passenger.name} booked on Flight {self.fl_num}.")
        

flight = Flight("AI202", 2)
flight.book(Passenger("Alice"))
flight.book(Passenger("Bob"))
flight.book(Passenger("Charlie"))
flight.book(Passenger("Charlie"))
```
